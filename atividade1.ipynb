```python
import pandas 

df = pandas.read_csv("C:/Users/Aluno.SantaLuzia/Documents/Elaine/uc2/bases/01.amazon_sales_dataset.csv")

descontos = df['discount']

descontoMedia = descontos.mean()
descontoMediana = descontos.median()
descontoMax = descontos.max()
descontoMin = descontos.min()
amplitude = descontoMax - descontoMin
q1 = descontos.quantile(0.25)
q3 = descontos.quantile(0.75)
assimetria = descontos.skew()
curtose = descontos.kurtosis()
delta = {
    'diferenca' : descontoMedia - descontoMediana,
    'razao' : descontoMedia / descontoMediana,
}

print('Media: ', descontoMedia)
print('Mediana: ', descontoMediana)
print('Desconto minimo: ', descontoMin)
print('Q1 (25%): ', q1)
print('Q3 (75%): ', q3)
print('Desconto máximo: ', descontoMax)
print('Amplitude(máx - min): ', amplitude)
print('Assimetria: ', assimetria)
print('Curtose: ', curtose)
print(f'Delta: {delta['diferenca']}, {delta['razao']}')
```

Os valores de média e mediana próximos um do outro indicam que os dados estão equilibrados/bem distribuídos;
A assimetria quase zero indica que o gráfico é quase simétrico;
A curtose indica que as caldas são leves e o pico do gráfico não é acentuado, logo os dados estão próximos da média;
